```python
market=[]
for i in range(0,dataset.shape[0]):
    cus=[]
    for j in dataset.columns :
        if type(dataset[j][i])==str:
            cus.append(dataset[j][i])
    market.append(cus)
```

```python
market[:10]
```

```python
l=[]
for i in market:
    for j in i :
        l.append(j)
```

```python
l[:10]
```

```python
import collections
```

```python
p = collections.Counter(l)
```

```python
d = {"Item Name":p.keys(),"values":p.values()}
```

```python
pd.DataFrame(d).sort_values(by=["values"],ascending=False).head(10)
```

```python
from mlxtend.preprocessing.transactionencoder import TransactionEncoder
```

```python
tr = TransactionEncoder()
tr.fit(market)
```

```python
df = pd.DataFrame(tr.transform(market),columns=tr.columns_)
```

```python
from mlxtend.frequent_patterns import apriori
```

```python
apriori(df, min_support=0.05,use_colnames=True,max_len=3).sort_values(by=["support"]).head(10)
```

```python
from mlxtend.frequent_patterns import fpgrowth
```

```python
fpgrowth(df,min_support=0.05,use_colnames=True,max_len=3).sort_values(by=["support"]).head(10)
```
